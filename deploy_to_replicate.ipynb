Run each cell in order. This will build and push your model to Replicate.

```
# Step 1: Free up disk space and check availability
!rm -rf /opt/hostedtoolcache /usr/local/lib/android /usr/share/dotnet /opt/conda 2>/dev/null
!apt-get clean 2>/dev/null
!df -h /
```

```
# Step 2: Install and start Docker
!curl -fsSL https://get.docker.com | sh

import subprocess, time

# Start dockerd in background using subprocess (not !)
proc = subprocess.Popen(
    ["sudo", "dockerd"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

for i in range(20):
    time.sleep(5)
    r = subprocess.run(["docker", "info"], capture_output=True)
    if r.returncode == 0:
        print("✅ Docker is running!")
        break
    print(f"Waiting for Docker... ({(i+1)*5}s)")
else:
    # Show error output for debugging
    r = subprocess.run(["docker", "info"], capture_output=True, text=True)
    print(r.stderr)
    raise RuntimeError("Docker failed to start after 100s")

!docker ps
```
